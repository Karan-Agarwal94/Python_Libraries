## Section 1: Setup & Creating a Dateset

````python
import pandas as pd
import numpy as np

# Messy customer dataset — intentional inconsistencies for cleaning practice
data = {
    "customer_id": [101, 102, 103, 104, 105, 106],
    "full_name": ["  alice johnson ", "BOB SMITH", "Charlie   Brown", "diana prince  ", "  EVAN WRIGHT", "Fiona green"],
    "email": ["Alice.Johnson@Gmail.COM", "bob.smith@yahoo.com", "CHARLIE@OUTLOOK.COM", "diana@gmail.com", "evan@HOTMAIL.COM", "fiona@gmail.com"],
    "phone": ["98-765-4321", "9123456780", "  91-234-56789  ", "9000011111", "88-888-8888", "7777777777"],
    "city": ["  Mumbai", "delhi", "BANGALORE", "Chennai  ", "hyderabad", "  Pune  "],
    "product_bought": ["Laptop Pro 15", "wireless MOUSE", "USB-C Hub (3-port)", "laptop pro 15", "  Keyboard  ", "wireless mouse"],
    "review": ["Great product!!", "average...not bad", "LOVED IT :)", "good, but expensive.", "  okay okay  ", "Best Buy Ever!!!"]
}

df = pd.DataFrame(data)
df
````

## Section 2: Basic Cleaning

````python
# --- Section 2: Basic String Cleaning ---

# 2.1 Strip leading/trailing whitespace
df["full_name"] = df["full_name"].str.strip()
df["city"] = df["city"].str.strip()
df["review"] = df["review"].str.strip()
df["phone"] = df["phone"].str.strip()

# 2.2 Fix casing
df["full_name"] = df["full_name"].str.title()   # Proper Name Format
df["city"] = df["city"].str.title()             # Proper City Format
df["email"] = df["email"].str.lower()           # emails are always lowercase
df["product_bought"] = df["product_bought"].str.strip().str.title()

print("After basic cleaning:\n")
df
````

## Section 3: Checking & Detecting

#### 3.1 Which customers used Gmail?

````python
gmail_users = df["email"].str.contains("gmail")
print("Gmail users (boolean):")
print(gmail_users)

print("\nFiltered — Gmail customers only:")
print(df[gmail_users][["customer_id", "full_name", "email"]])
````

#### 3.2 Which names start with a specific letter?

````python
print("\nNames starting with 'A':")
print(df[df["full_name"].str.startswith("A")]["full_name"])
````

#### 3.3 Which reviews end with '!'?

````python
print("\nEnthusiastic reviews (ending with '!'):")
print(df[df["review"].str.endswith("!")][["full_name", "review"]])
````

#### 3.4 Are phone numbers all digits?

````python
print("\nIs phone all digits?")
print(df["phone"].str.isdigit())
````

## Section 4: Extracting

#### 4.1 Extract first name and last name

````python
df['first_name'] = df['full_name'].str.split().str[0]
df['last_name'] = df['full_name'].str.split().str[1]
print("Full Names splitted in First and Last Name: \n")
print(df[['full_name', 'first_name', 'last_name']])
````

#### 4.2 Extracting Email Domains ( Everything after '@')

````python
df['email_domain'] = df['email'].str.split('@').str[1]
print("\nEmail domains:")
print(df[["email", "email_domain"]])
````

#### 4.3 Extracting first 3 characters of city as a region code

````python
df['city_code'] = df['city'].str[:3].str.upper()

print("\nCity codes:")
print(df[["city", "city_code"]])
````

#### 4.4 Extract digits only from phone using regex

````python
df["phone_digits"] = df["phone"].str.findall(r"\d+").str.join("")
print("\nFirst digit group extracted from phone:")
print(df[["phone", "phone_digits"]])
````

## Section 5: Replacing and Removing

#### 5.1 Fixing Phone Numbers ( Removing all hyphens)

````python
df["phone_clean"] = df['phone'].str.replace("-", "")
print(df[["phone", "phone_clean"]])
````

##### Important Note:
In `.str.replace()`, we have one more parameter provided by the pandas, which is `regex=False` ( Default behaviour is False in modern Pandas).

So, we can explicitely use `regex=True` and customize the data with more options. 

Following are the special characters that trigger difference:
> 
. * + ? ^ $ | \ [ ] ( ) { }

Sine your pattern is just a hyphen "-" (not special in regex), both approaches work identically.

#### 5.2 Fixing the USB-C product name using `str.replace()`

````python
df["product_bought"] = df["product_bought"].str.replace("Usb-C", "USB-C")
print(df["product_bought"])
````

#### 5.3 Remove punctuation from reviews using `regex=True`

````python
df["review_clean"] = df["review"].str.replace(r"[^\w\s]", "", regex=True).str.title()
print(df[["review", "review_clean"]])
````

##### Observation:

The regex pattern: r"[^\w\s]"

- [^...] = Negated character class (match anything NOT listed)
- \w = Word characters (a-z, A-Z, 0-9, _)
- \s = Whitespace (spaces, tabs, newlines)

So [^\w\s] means: match any character that is neither a word character nor whitespace.

#### 5.4 Replace multiple spaces inside strings (e.g. "Charlie   Brown")

````python
df["full_name"] = df["full_name"].str.replace(r"\s+", " ", regex=True).str.strip()

print(df["full_name"])
````

## Section 6: Splitting and Joining

#### 6.1 Split email into username and domain as separate columns

````python
email_split = df["email"].str.split("@", expand=True)
email_split.columns = ["email_username", "email_domain_full"]
print(email_split)
````

##### Note:
`expand= True`: Returns a DataFrame where each split piece become a separate column.

`expand= False` (default behaviour): Returns a Series of Lists.

#### 6.2 Rejoin first_name and last_name with a comma (Last, First format)

````python
df["formal_name"] = df["last_name"] + ", " + df["first_name"]

print(df[["full_name", "formal_name"]])
````

#### 6.3 Concatenate city and email_domain into a contact label using `str.cat()`

````python
df["contact_label"] = df["city"].str.cat(df["email_domain"], sep=" | ")

print(df[["full_name", "city", "email_domain", "contact_label"]])
````

## Section 7: String Length and Counting 

#### 7.1 Length of each customer's full name

````python
df["name_length"] = df["full_name"].str.len()

print(df[["full_name", "name_length"]])
````

#### 7.2 Word count in each review

````python
df["review_word_count"] = df["review"].str.split().str.len()

print(df[["review", "review_word_count"]])
````

#### 7.3 Count occurrences of a specific character

````python
df["exclamation_count"] = df["review"].str.count(r"!")

print(df[["review", "exclamation_count"]])
````

#### 7.4 Who wrote the longest and shortest review?

````python
print("\nLongest review:")
print(df.loc[df["review_word_count"].idxmax(), ["full_name", "review"]])

print("\nShortest review:")
print(df.loc[df["review_word_count"].idxmin(), ["full_name", "review"]])
````

##### Observation:
`idxmax/idxmin` picks the FIRST occurrence when there are ties

````python
# Better approach — show ALL longest and shortest reviews

max_words = df["review_word_count"].max()
min_words = df["review_word_count"].min()

print(f"All longest reviews ({max_words} words):")
print(df[df["review_word_count"] == max_words][["full_name", "review"]])

print(f"\nAll shortest reviews ({min_words} words):")
print(df[df["review_word_count"] == min_words][["full_name", "review"]])
````

## Section 8: Real World Mini Task

````python
data = {
    "emp_id": [201, 202, 203, 204, 205, 206, 207],
    "emp_name": ["  ravi KUMAR ", "Sneha Mehta", "ARJUN das", "priya SHARMA  ", "  Karan Singh", "NEHA joshi  ", "vikram  Bhat"],
    "department": ["  engineering", "MARKETING", "Finance  ", "engineering", "  MARKETING  ", "finance", "Engineering"],
    "email": ["Ravi.Kumar@COMPANY.COM", "sneha.mehta@company.com", "ARJUN@COMPANY.COM", "priya@company.com", "Karan.Singh@Company.Com", "neha@COMPANY.COM", "vikram@company.com"],
    "phone": ["98-765-43210", "9123456780", "  88-888-88888  ", "9000011111", "77-777-77777", "9876543210", "  9111122222"],
    "salary": ["85,000", "72000", "1,10,000", "95,000", "68000", "1,20,000", "76,000"],
    "join_date": ["15-Aug-2019", "2020/03/22", "March 10, 2021", "18-Jan-2018", "2022/07/05", "June 1, 2020", "09-Nov-2017"],
    "performance": ["  excellent  ", "GOOD", "Average", "excellent", "  POOR  ", "good", "  Average  "]
}

df = pd.DataFrame(data)
df
````

#### Tasks to be done:
- emp_name — Strip whitespace, fix to Title Case, collapse internal extra spaces, then split into first_name and last_name
- department — Strip, Title Case, then verify: how many unique departments are there?
- email — Lowercase, then extract email_username (before @) and email_provider (after @) as separate columns
- phone — Strip whitespace, remove hyphens, check which entries are all digits using str.isdigit()
- salary — Remove all commas, convert to integer, then find the highest and lowest paid employee
- join_date — This one is tricky — three different date formats are mixed. Just use pd.to_datetime(df["join_date"], infer_datetime_format=True) and see what happens
- performance — Strip, lowercase, then use str.replace() to create a numeric perf_score column: excellent→4, good→3, average→2, poor→1

#### 8.1 `emp_name`, `first_name`, `last_name`

````python
# Creating an empty DataFrame 
clean_df = pd.DataFrame()

clean_df['emp_name'] = df['emp_name'].str.strip().str.title().str.replace(r"\s+", " ", regex=True)

clean_df['first_name'] = clean_df['emp_name'].str.split(" ").str[0]
clean_df['last_name'] = clean_df['emp_name'].str.split(" ").str[1]

clean_df[['emp_name', 'first_name', 'last_name']]
````

#### 8.2 `department` & Count of Unique Departments

````python
clean_df['department'] = df['department'].str.strip().str.title()
print(clean_df['department'])
````

````python
# Counting the employees inside a department using `value_counts()`
clean_df['department'].value_counts()
````

````python
# Counting the number of unique values inside the column department using `nunique()`
clean_df['department'].nunique()
````

````python
# Displaying the unique values inside a colunn using `unique()` 
clean_df['department'].unique()
````

#### 8.3 `email`, `email_username`, `email_provider`

````python
clean_df['email'] = df['email'].str.strip().str.lower()

clean_df['email_username'] = clean_df['email'].str.split('@').str[0]
clean_df['email_provider'] = clean_df['email'].str.split('@').str[1]
````

#### 8.4 `phone`

````python
# Checking which phone numbers are all digit in messy dataframe
df['phone'].str.isdigit()
````

````python
# cleaning phone numbers
clean_df['phone'] = df['phone'].str.strip().str.replace('-', '') 
clean_df['phone']
````

````python
# Rechecking phone whether all values are digits or not
clean_df['phone'].str.isdigit()
````

#### 8.5 `salary` and Highest & Lowest paid employees

````python
# Checking current value in messy data
df['salary'].str.isdigit()
````

````python
clean_df['salary'] = df['salary'].str.replace(r"[^\w]", "", regex=True)

clean_df['salary'] = pd.to_numeric(clean_df['salary'])

# sanity check
print(clean_df['salary'].dtype)
````

````python
# Highest_paid employee
highest_salary = clean_df['salary'].max()

print(clean_df[['emp_name', 'department', 'salary']][clean_df['salary'] == highest_salary])

# also print(clean_df[['emp_name', 'department', 'salary']][clean_df['salary'] == clean_df['salary'].max()])
````

````python
# Lowest_paid employee
Lowest_salary = clean_df['salary'].min()

print(clean_df[['emp_name', 'department', 'salary']][clean_df['salary'] == Lowest_salary])

# also print(clean_df[['emp_name', 'department', 'salary']][clean_df['salary'] == clean_df['salary'].min()])
````

#### 8.6 `join_date`

````python
# Checking current datatype of join_date column
print(df['join_date'])
````

````python
from dateutil import parser

# This handles ALL formats reliably
clean_df['join_date'] = df['join_date'].apply(
    lambda x: parser.parse(x) if pd.notna(x) else pd.NaT
)

clean_df['join_date']
````

#### Interpretation:
```
clean_df['join_date'] = pd.to_datetime(
    df['join_date'],
    errors='coerce'
)

- pd.to_datetime() is vectorized, faster and cleaner but couldn't handle messy dates.
- Therefore, dateutil.parser is being used here.
- In the same way, `pd.to_datetime(df["join_date"], infer_datetime_format=True)` also couldn't handle this messy data and converted most of the dates into NaT.

#### 8.7 `performance` and `perf_score`:

````python
clean_df['performance'] = df['performance'].str.strip().str.lower()
````

````python
# Creating perf_score based on performance column
clean_df['perf_score'] = clean_df['performance'].map({
                                    'excellent': 4,
                                    'good'     : 3,
                                    'average'  : 2,
                                    'poor'     : 1
}).astype('Int64')
````

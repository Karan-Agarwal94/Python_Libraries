## Section 1: Datatime Basics

Time series data is any data indexed or associated with timestamps.

Pandas has first-class support for time series via `datetime64`, `Timestamp`, and `DatetimeIndex`.

#### 1.1 Creating a single Timestamp through `pd.Timestamp()`

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Creating a single Timestamp
ts = pd.Timestamp("2026-05-02")
print(ts)
print(type(ts))
print("Year:", ts.year, "| Month:", ts.month, "| Day:", ts.day)
```

#### 1.2 `pd.to_datetime()` -- Converting strings to datetime

```python
# Converting strings to datetime
dates = ["2024-01-01", "2024-02-15", "2024-03-30"]
dt_series = pd.to_datetime(dates)
print(dt_series)
print("dtype:", dt_series.dtype)
```

#### `pd.to_datetime()` -- Coverting DataFrame column to Datetime

```python
# Converting a DataFrame column to datetime
df = pd.DataFrame({
    "date": ["2024-01-01", "2024-01-02", "2024-01-03"],
    "sales": [200, 350, 410]
})

df["date"] = pd.to_datetime(df["date"])
df = df.set_index("date")     # Setting `date` column as index

print(df.dtypes)
print(df.head())
```

## Section 2: DateTime Index

When a datetime column is set as the index, Pandas creates a `DatetimeIndex`.

This unlocks powerful time-based indexing, slicing, and property access.

#### 2.1 Inspecting Datetime Index Properties

```python
df = pd.DataFrame({
    "date": pd.date_range(start="2024-01-01", periods=6, freq="ME"),
    "sales": [200, 350, 410, 290, 500, 620]
})
df = df.set_index("date")
print(df)
print()

# Inspect the index
print("Index type :", type(df.index))
print("Index dtype:", df.index.dtype)
```

#### 2.2 Extracting Date components from the Index

```python
# Date component properties on a DatetimeIndex
print("Year  :", df.index.year.tolist())
print("Month :", df.index.month.tolist())
print("Day   :", df.index.day.tolist())
print("Weekday:", df.index.day_name().tolist())
```

#### 2.3 Filtering rows using index properties

```python
# Filter rows where month is greater than 3 (i.e., April onwards)
q2_onwards = df[df.index.month > 3]
print(q2_onwards)
```

## Section 3: `pd.date_range()` -- Date Range

`pd.date_range()` generates a sequence of evenly-spaced datetime values.

**Key parameters**: `start`, `end`, `periods`, `freq`

**Common frequencies**: D (day), W (week), ME (month-end), MS (month-start), H (hour), min (minute)

#### 3.1 Basic Date Range

```python
# Daily range — using start + periods
daily = pd.date_range(start="2024-01-01", periods=7, freq="D")
print("Daily:\n", daily)

# Weekly range — using start + end
weekly = pd.date_range(start="2024-01-01", end="2024-03-31", freq="W")
print("\nWeekly:\n", weekly)
```

#### 3.2 Month Start vs Month End

```python
# Month-start (MS) vs Month-end (ME)
ms = pd.date_range(start="2024-01-01", periods=4, freq="MS")
me = pd.date_range(start="2024-01-01", periods=4, freq="ME")

print("Month-Start:", ms.tolist())
print("Month-End  :", me.tolist())
```

#### 3.3 Hourly Range

```python
# Hourly — useful for IoT / sensor / log data
hourly = pd.date_range(start="2024-01-01 08:00", periods=6, freq="h")
print("Hourly:\n", hourly)
```

#### 3.4 Using `pd.date_range()` to build a DataFrame

```python
# Simulate daily sales data for Jan 2024
np.random.seed(42)

date_idx = pd.date_range(start="2024-01-01", periods=31, freq="D")

sales_df = pd.DataFrame({
    "sales": np.random.randint(100, 600, size=31)
}, index=date_idx)

sales_df.index.name = "date"
print(sales_df.head(7))
print("Shape:", sales_df.shape)
```

## Section 4: Indexing and Slicing Time Series

With a DatetimeIndex, Pandas supports powerful partial string indexing —

you can slice by year, month, or exact date using `.loc[]`.

#### 4.1 Selecting a single Date

```python
# Exact date lookup
print(sales_df.loc["2024-01-05"])
```

#### 4.2 Slice a Date Range

```python
# Slice a range of dates using partial strings
print(sales_df.loc["2024-01-10":"2024-01-15"])
```

#### 4.3 Partial string Indexing by Month

```python
# Select all rows for a given month (entire January)
print(sales_df.loc["2024-01"].sample(5))
```

#### 4.4 First and Last N rows using Time Aware Methods

```python
# First 5 days and last 5 days
print("First 5 days:")
print(sales_df.head(5))

print("\nLast 5 days:")
print(sales_df.tail(5))
```

#### 4.5 Boolean Filtering with Index

```python
# Sales only on weekdays (Mon=0 to Fri=4)
weekday_sales = sales_df[sales_df.index.dayofweek < 5]
weekend_sales = sales_df[sales_df.index.dayofweek >= 5]

print("Weekday rows:", len(weekday_sales))
print("Weekend rows:", len(weekend_sales))
print("\nWeekend sales:\n", weekend_sales)
```

## Section 5: Resampling

Resampling changes the frequency of time series data.

**Downsampling:** higher freq → lower freq (e.g., daily → weekly). Requires aggregation.

**Upsampling:** lower freq → higher freq (e.g., daily → hourly). Requires filling strategy.

#### 5.1 Downsample: Daily --> Weekly

```python
# Weekly total sales (sum)
weekly_sales = sales_df.resample("W").sum()
print("Weekly Total Sales:")
print(weekly_sales)
```

#### 5.2 Downsample with Multiple Aggregations

```python
# Weekly summary — min, max, mean, sum
weekly_summary = sales_df.resample("W").agg(
    min_sales=("sales", "min"),
    max_sales=("sales", "max"),
    avg_sales=("sales", "mean"),
    total_sales=("sales", "sum")
)
print(weekly_summary)
```

#### 5.3 Downsample: Daily ---> Monthly 

```python
# Monthly aggregation
monthly = sales_df.resample("ME").agg(["sum", "mean", "max"])
print(monthly)
```

#### 5.4 Upsample: Daily ---> 12 Hourly

```python
# Upsample to 12-hour frequency — creates NaN for new timestamps
upsampled = sales_df.resample("12h").asfreq()
print("Upsampled (first 8 rows):")
print(upsampled.head(8))
```

#### 5.5 Filling NaNs after Unsampling

```python
# Forward fill — carry last known value forward
ffill_df = sales_df.resample("12h").ffill()
print("Forward Filled (first 8 rows):")
print(ffill_df.head(8))
```

## Section 6: Rolling and Expanding Windows

**Rolling:** computes over a fixed-size sliding window (e.g., last 7 days).

**Expanding:** computes over all data from the start up to the current row.

Both are used heavily for moving averages, trend smoothing, and cumulative metrics.

#### 6.1 Rolling Mean: --> 7-day Moving Average

```python
# 7-day rolling average — first 6 rows will be NaN (insufficient window)
sales_df["rolling_7d_mean"] = sales_df["sales"].rolling(window=7).mean()
print(sales_df.head(10))
```

#### 6.2 Rollwing with `min_periods`

```python
# min_periods=1 — start computing even with fewer than 7 rows
sales_df["rolling_7d_minp"] = sales_df["sales"].rolling(window=7, min_periods=1).mean()
print(sales_df.head(10))
```

#### 6.3 Rolling `min` and `max`

```python
# Rolling 3-day min and max
sales_df["rolling_3d_min"] = sales_df["sales"].rolling(window=3).min()
sales_df["rolling_3d_max"] = sales_df["sales"].rolling(window=3).max()
print(sales_df[["sales", "rolling_3d_min", "rolling_3d_max"]].head(12))
```

#### 6.4 Expanding Window -- `df.expanding()`

```python
# Expanding mean — cumulative average from day 1 up to current row
sales_df["expanding_mean"] = sales_df["sales"].expanding().mean()
print(sales_df[["sales", "expanding_mean"]].round(2).head(10))
```

#### 6.5 Cumulative Sum / Running Sum -- `.cumsum()`

```python
# Cumulative sales total
sales_df["cumulative_sales"] = sales_df["sales"].cumsum()
print(sales_df[["sales", "cumulative_sales"]].tail(7))
```

## Section 7: Shifting & Lag -- `.shift()` & `.diff()`

.shift(n): moves data n periods forward (positive) or backward (negative).

.diff(n): computes the difference between current and n periods ago.

**Used for:**
- lag features,
- day-over-day change,
- % growth calculations.

#### 7.1 `.shift()` to create Lag features

```python
# Lag 1 — yesterday's sales alongside today's
sales_df["lag_1"] = sales_df["sales"].shift(1)
sales_df["lag_2"] = sales_df["sales"].shift(2)
print(sales_df[["sales", "lag_1", "lag_2"]].head(8))
```

#### 7.2 `.diff()` for day over day Change

```python
# Absolute change vs previous day
sales_df["daily_change"] = sales_df["sales"].diff(1)
print(sales_df[["sales", "daily_change"]].head(8))
```

#### 7.3 Percentage change with `.pct_change()`

```python
# % change vs previous day
sales_df["pct_change"] = sales_df["sales"].pct_change() * 100
print(sales_df[["sales", "pct_change"]].round(2).head(8))
```

#### 7.4 Week over Week change using `.shift(7)`

```python
# Compare today vs same day last week
sales_df["wow_change"] = sales_df["sales"] - sales_df["sales"].shift(7)
print(sales_df[["sales", "wow_change"]].tail(10))
```

## Section 8: Time Zones — `tz_localize()` & `tz_convert()`

By default, Pandas datetime objects are timezone-naive.

tz_localize(): assigns a timezone to a naive datetime (declares "this is UTC" etc.)

tz_convert(): converts from one timezone to another (requires already localized data)

**Relevant for:**
- APIs,
- logs,
- financial data,
- and any global data pipelines.

#### 8.1 Check Timezone awareness

```python
# Our sales_df index is timezone-naive
print("Timezone:", sales_df.index.tz)
```

#### 8.2 `tz_localize()`: assign a timezone

```python
# Localize to UTC first
sales_utc = sales_df[["sales"]].copy()
sales_utc.index = sales_utc.index.tz_localize("UTC")
print("Timezone after localize:", sales_utc.index.tz)
print(sales_utc.head(3))
```

#### 8.3 `tz_convert()`: Convert to another timezone

```python
# Convert UTC → India Standard Time
sales_ist = sales_utc.copy()
sales_ist.index = sales_ist.index.tz_convert("Asia/Kolkata")
print("Timezone after convert:", sales_ist.index.tz)
print(sales_ist.head(3))

# Convert UTC → US Eastern
sales_us = sales_utc.copy()
sales_us.index = sales_us.index.tz_convert("US/Eastern")
print("\nUS Eastern:")
print(sales_us.head(3))
```

#### 8.4 Side by Side Comparison

```python
# Compare the same instant across timezones
compare = pd.DataFrame({
    "UTC"    : sales_utc["sales"].head(3).values,
    "IST"    : sales_ist["sales"].head(3).values,
    "US_EST" : sales_us["sales"].head(3).values,
}, index=sales_utc.head(3).index)

print(compare)
print("\nNote: sales values are identical — only the index timezone label changes.")
```

## Section 9: Mini Project — End-to-End Time Series Analysis

Scenario:
- You are a data analyst at a retail company.
- You have 3 months of daily sales data (Jan–Mar 2024).
- Task: Clean, enrich, resample, and summarize the data for a business report.

#### 9.1 Building The Dataset

```python
np.random.seed(99)

date_idx = pd.date_range(start="2024-01-01", end="2024-03-31", freq="D")

ts = pd.DataFrame({
    "sales"    : np.random.randint(150, 750, size=len(date_idx)),
    "returns"  : np.random.randint(0,   50, size=len(date_idx))
}, index=date_idx)

ts.index.name = "date"
print("Shape:", ts.shape)
print(ts.head(5))
```

#### 9.2 Enriching with time features

```python
ts["month"]      = ts.index.month_name()
ts["weekday"]    = ts.index.day_name()
ts["is_weekend"] = ts.index.dayofweek >= 5
ts["net_sales"]  = ts["sales"] - ts["returns"]
print(ts.sample(10))
```

#### 9.3 Weekly Summary Report

```python
weekly_report = ts.resample("W").agg(
    total_sales   = ("sales",     "sum"),
    total_returns = ("returns",   "sum"),
    net_sales     = ("net_sales", "sum"),
    avg_daily     = ("sales",     "mean"),
    best_day      = ("sales",     "max"),
    worst_day     = ("sales",     "min")
).round(2)

print("Weekly Report:")
print(weekly_report)
```

#### 9.4 Last 7 days of each month with rolling trend

```python
ts["rolling_7d"] = ts["net_sales"].rolling(7).mean().round(2)

# Show last 7 days of each month with rolling trend
for month, group in ts[["net_sales", "rolling_7d"]].resample("ME"):
    print(f"\n── {month.strftime('%B %Y')} (last 7 days) ──")
    print(group.tail(7).to_string())
```

#### 9.5 Weekday vs Weekend Average Sales

```python
comparison = ts.groupby("is_weekend")["sales"].agg(["mean", "sum", "count"]).round(2)
comparison.index = ["Weekday", "Weekend"]
print("Weekday vs Weekend Sales:")
print(comparison)
```

#### Key Insights
- Dataset spans Jan–Mar 2024 (91 days)
- Net sales = sales minus returns, tracked daily
- Weekly report shows best and average day per week
- Rolling 7-day mean smooths daily noise for trend visibility
- Weekend vs Weekday breakdown reveals demand patterns
